## b) Diseño y entrenamiento de un modelo con embeddings

```python
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# --- Target ---
y_train = (train["income"] == ">50K").astype(np.float32).values
y_val   = (val["income"] == ">50K").astype(np.float32).values

# --- Variables numéricas ---
num_cols = ["age", "capital-gain", "capital-loss", "hours-per-week"]
scaler = StandardScaler().fit(train[num_cols])
X_num_train = scaler.transform(train[num_cols]).astype(np.float32)
X_num_val   = scaler.transform(val[num_cols]).astype(np.float32)

# --- Variable ordinal: education ---
education_order = ["Preschool", "1st-4th", "5th-6th", "7th-8th", "9th", "10th",
                    "11th", "12th", "HS-grad", "Some-college", "Assoc-voc",
                    "Assoc-acdm", "Bachelors", "Masters", "Prof-school", "Doctorate"]
edu_map = {cat: i for i, cat in enumerate(education_order)}
X_edu_train = train["education"].map(edu_map).values.astype(np.float32).reshape(-1, 1)
X_edu_val   = val["education"].map(edu_map).values.astype(np.float32).reshape(-1, 1)
edu_scaler = StandardScaler().fit(X_edu_train) # se normaliza la ordinal para que quede en escala similar al resto
X_edu_train = edu_scaler.transform(X_edu_train).astype(np.float32)
X_edu_val   = edu_scaler.transform(X_edu_val).astype(np.float32)

# --- One-Hot Encoding: baja cardinalidad ---
onehot_cols = ["workclass", "marital-status", "relationship", "race", "sex"]
ohe = OneHotEncoder(handle_unknown="ignore", sparse_output=False).fit(train[onehot_cols])
X_ohe_train = ohe.transform(train[onehot_cols]).astype(np.float32)
X_ohe_val   = ohe.transform(val[onehot_cols]).astype(np.float32)

# --- Embeddings: alta cardinalidad ---
emb_cols = ["occupation", "native-country", "skill-profile"]

emb_maps = {}       # categoria -> indice, 0 reservado para "unknown"
emb_cardinalities = {}
for col in emb_cols:
    cats = sorted(train[col].unique())
    emb_maps[col] = {cat: i + 1 for i, cat in enumerate(cats)}  # 0 = unknown
    emb_cardinalities[col] = len(cats) + 1  # +1 por el índice "unknown"

def encode_emb(df, col):
    return df[col].map(emb_maps[col]).fillna(0).astype(np.int64).values

X_emb_train = {col: encode_emb(train, col) for col in emb_cols}
X_emb_val   = {col: encode_emb(val, col) for col in emb_cols}  # categorías nuevas -> 0
```

```python
class AdultDataset(Dataset):
    def __init__(self, X_num, X_ohe, X_emb, y):
        self.X_num = torch.tensor(np.hstack([X_num, X_ohe]), dtype=torch.float32)
        self.X_emb = {col: torch.tensor(v, dtype=torch.long) for col, v in X_emb.items()}
        self.y = torch.tensor(y, dtype=torch.float32)

    def __len__(self):
        return len(self.y)

    def __getitem__(self, idx):
        emb_idx = {col: v[idx] for col, v in self.X_emb.items()}
        return self.X_num[idx], emb_idx, self.y[idx]

# Concatenación
X_dense_train = np.hstack([X_num_train, X_edu_train, X_ohe_train])
X_dense_val   = np.hstack([X_num_val, X_edu_val, X_ohe_val])

train_ds = AdultDataset(X_dense_train, np.empty((len(X_dense_train), 0)), X_emb_train, y_train)
val_ds   = AdultDataset(X_dense_val, np.empty((len(X_dense_val), 0)), X_emb_val, y_val)
```

```python
def embedding_dim(n_categories):
    return min(50, round(1.6 * n_categories ** 0.56))

emb_dims = {col: embedding_dim(emb_cardinalities[col]) for col in emb_cols}
print(emb_dims)
```

```python
import random
seed = 42
random.seed(seed); np.random.seed(seed); torch.manual_seed(seed); torch.cuda.manual_seed_all(seed)

class EmbeddingMLP(nn.Module):
    def __init__(self, n_dense_features, emb_cardinalities, emb_dims, hidden_dims=(256, 128, 64), p=0.3):
        super().__init__()
        self.emb_cols = list(emb_cardinalities.keys())

        self.embeddings = nn.ModuleDict({
            col: nn.Embedding(emb_cardinalities[col], emb_dims[col])
            for col in self.emb_cols
        })
        for col in self.emb_cols:
            nn.init.normal_(self.embeddings[col].weight, std=0.01)

        total_emb_dim = sum(emb_dims.values())
        input_dim = n_dense_features + total_emb_dim

        self.dropout = nn.Dropout(p)

        layers = []
        prev_dim = input_dim
        for h in hidden_dims:
            layers += [nn.Linear(prev_dim, h), nn.LeakyReLU(), nn.Dropout(p)]
            prev_dim = h
        layers += [nn.Linear(prev_dim, 1)]
        self.mlp = nn.Sequential(*layers)

    def forward(self, x_dense, emb_idx):
        emb_out = [self.embeddings[col](emb_idx[col]) for col in self.emb_cols]
        x = torch.cat([x_dense] + emb_out, dim=1)
        x = self.dropout(x)
        return self.mlp(x).squeeze(1)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
n_dense = X_dense_train.shape[1]
model = EmbeddingMLP(n_dense, emb_cardinalities, emb_dims, hidden_dims=(256, 128, 64), p=0.3).to(device)
criterion = nn.BCEWithLogitsLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
```

```python
from sklearn.metrics import accuracy_score, f1_score, classification_report, confusion_matrix

device = "cuda" if torch.cuda.is_available() else "cpu"
model.to(device)

criterion = nn.BCEWithLogitsLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

train_loader = DataLoader(train_ds, batch_size=256, shuffle=True)
val_loader   = DataLoader(val_ds, batch_size=512, shuffle=False)

def run_epoch(loader, train_mode):
    model.train() if train_mode else model.eval()
    all_preds, all_targets = [], []
    total_loss = 0.0

    with torch.set_grad_enabled(train_mode):
        for x_dense, emb_idx, y in loader:
            x_dense = x_dense.to(device)
            emb_idx = {k: v.to(device) for k, v in emb_idx.items()}
            y = y.to(device)

            logits = model(x_dense, emb_idx)
            loss = criterion(logits, y)

            if train_mode:
                optimizer.zero_grad()
                loss.backward()
                optimizer.step()

            total_loss += loss.item() * len(y)
            preds = (torch.sigmoid(logits) > 0.5).float()
            all_preds.append(preds.cpu().numpy())
            all_targets.append(y.cpu().numpy())

    all_preds = np.concatenate(all_preds)
    all_targets = np.concatenate(all_targets)
    acc = accuracy_score(all_targets, all_preds)
    f1_macro = f1_score(all_targets, all_preds, average="macro")
    return total_loss / len(loader.dataset), acc, f1_macro

n_epochs = 30
history = {"train_acc": [], "val_acc": [], "train_f1": [], "val_f1": []}

for epoch in range(n_epochs):
    train_loss, train_acc, train_f1 = run_epoch(train_loader, train_mode=True)
    val_loss, val_acc, val_f1 = run_epoch(val_loader, train_mode=False)

    history["train_acc"].append(train_acc)
    history["val_acc"].append(val_acc)
    history["train_f1"].append(train_f1)
    history["val_f1"].append(val_f1)

    print(f"Epoch {epoch+1}/{n_epochs} - train_loss={train_loss:.4f} val_loss={val_loss:.4f} "
          f"train_acc={train_acc:.4f} val_acc={val_acc:.4f} train_f1={train_f1:.4f} val_f1={val_f1:.4f}")
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(history["train_acc"], label="Train")
axes[0].plot(history["val_acc"], label="Val")
axes[0].set_title("Accuracy vs Epoch")
axes[0].set_xlabel("Epoch"); axes[0].set_ylabel("Accuracy"); axes[0].legend()

axes[1].plot(history["train_f1"], label="Train")
axes[1].plot(history["val_f1"], label="Val")
axes[1].set_title("F1 macro vs Epoch")
axes[1].set_xlabel("Epoch"); axes[1].set_ylabel("F1 macro"); axes[1].legend()

plt.tight_layout()
plt.show()
```

```python
# obtengo predicciones finales sobre val
model.eval()
all_preds, all_targets = [], []
with torch.no_grad():
    for x_dense, emb_idx, y in val_loader:
        x_dense = x_dense.to(device)
        emb_idx = {k: v.to(device) for k, v in emb_idx.items()}
        logits = model(x_dense, emb_idx)
        preds = (torch.sigmoid(logits) > 0.5).float()
        all_preds.append(preds.cpu().numpy())
        all_targets.append(y.numpy())

all_preds = np.concatenate(all_preds)
all_targets = np.concatenate(all_targets)

print(classification_report(all_targets, all_preds, target_names=["<=50K", ">50K"]))

cm_abs = confusion_matrix(all_targets, all_preds)
cm_norm = confusion_matrix(all_targets, all_preds, normalize="true")

fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))
sns.heatmap(cm_abs, annot=True, fmt="d", cmap="Blues",
            xticklabels=["<=50K", ">50K"], yticklabels=["<=50K", ">50K"], ax=axes[0])
axes[0].set_title("Matriz de confusión (absoluta)")
axes[0].set_xlabel("Predicho"); axes[0].set_ylabel("Real")

sns.heatmap(cm_norm, annot=True, fmt=".2f", cmap="Blues",
            xticklabels=["<=50K", ">50K"], yticklabels=["<=50K", ">50K"], ax=axes[1])
axes[1].set_title("Matriz de confusión (normalizada por fila)")
axes[1].set_xlabel("Predicho"); axes[1].set_ylabel("Real")

plt.tight_layout()
plt.show()
```

## c) Diseño y entrenamiento de un modelo sin embeddings

```python
# ahora las categóricas (excepto education, que sigue ordinal) van a One-Hot Encoding
onehot_cols_c = ["workclass", "marital-status", "relationship", "race", "sex","occupation", "native-country", "skill-profile"]

ohe_c = OneHotEncoder(handle_unknown="ignore", sparse_output=False).fit(train[onehot_cols_c])
X_ohe_train_c = ohe_c.transform(train[onehot_cols_c]).astype(np.float32)
X_ohe_val_c   = ohe_c.transform(val[onehot_cols_c]).astype(np.float32)

print("Dimensión one-hot:", X_ohe_train_c.shape[1])
```

```python
X_dense_train_c = np.hstack([X_num_train, X_edu_train, X_ohe_train_c])
X_dense_val_c   = np.hstack([X_num_val, X_edu_val, X_ohe_val_c])

print("Input total modelo c:", X_dense_train_c.shape[1])
print("Input total modelo b (dense + embeddings):", n_dense + sum(emb_dims.values()))
```

```python
class AdultDatasetNoEmb(Dataset):
    def __init__(self, X_dense, y):
        self.X = torch.tensor(X_dense, dtype=torch.float32)
        self.y = torch.tensor(y, dtype=torch.float32)

    def __len__(self):
        return len(self.y)

    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]

train_ds_c = AdultDatasetNoEmb(X_dense_train_c, y_train)
val_ds_c   = AdultDatasetNoEmb(X_dense_val_c, y_val)

train_loader_c = DataLoader(train_ds_c, batch_size=256, shuffle=True)
val_loader_c   = DataLoader(val_ds_c, batch_size=512, shuffle=False)
```

```python
class MLPNoEmbeddings(nn.Module):
    def __init__(self, input_dim, hidden_dims=(256, 128, 64)):
        super().__init__()
        layers = []
        prev_dim = input_dim
        for h in hidden_dims:
            layers += [nn.Linear(prev_dim, h), nn.LeakyReLU()]  # sin Dropout
            prev_dim = h
        layers += [nn.Linear(prev_dim, 1)]
        self.mlp = nn.Sequential(*layers)

    def forward(self, x):
        return self.mlp(x).squeeze(1)

input_dim_c = X_dense_train_c.shape[1]
model_c = MLPNoEmbeddings(input_dim_c, hidden_dims=(256, 128, 64)).to(device)
criterion_c = nn.BCEWithLogitsLoss()
optimizer_c = torch.optim.Adam(model_c.parameters(), lr=1e-3)
```

```python
def evaluate_c(loader):
    model_c.eval()
    all_preds, all_targets = [], []
    with torch.no_grad():
        for x, y in loader:
            x = x.to(device)
            logits = model_c(x)
            preds = (torch.sigmoid(logits) > 0.5).float().cpu().numpy()
            all_preds.append(preds)
            all_targets.append(y.numpy())
    all_preds = np.concatenate(all_preds)
    all_targets = np.concatenate(all_targets)
    acc = accuracy_score(all_targets, all_preds)
    f1_macro = f1_score(all_targets, all_preds, average="macro")
    return acc, f1_macro

epochs = 30
train_acc_hist_c, val_acc_hist_c = [], []
train_f1_hist_c, val_f1_hist_c = [], []

for epoch in range(1, epochs + 1):
    model_c.train()
    train_losses = []
    for x, y in train_loader_c:
        x, y = x.to(device), y.to(device)
        logits = model_c(x)
        loss = criterion_c(logits, y)

        optimizer_c.zero_grad()
        loss.backward()
        optimizer_c.step()
        train_losses.append(loss.item())

    train_acc, train_f1 = evaluate_c(train_loader_c)
    val_acc, val_f1 = evaluate_c(val_loader_c)

    train_acc_hist_c.append(train_acc)
    val_acc_hist_c.append(val_acc)
    train_f1_hist_c.append(train_f1)
    val_f1_hist_c.append(val_f1)

    print(f"Epoch {epoch} | train_loss={np.mean(train_losses):.4f} | "
          f"train_acc={train_acc:.4f} | val_acc={val_acc:.4f} | "
          f"train_f1={train_f1:.4f} | val_f1={val_f1:.4f}")
```

```python
plt.figure(figsize=(7,4))
plt.plot(range(1, epochs+1), train_acc_hist_c, marker="o", label="Train")
plt.plot(range(1, epochs+1), val_acc_hist_c, marker="o", label="Val")
plt.xlabel("Época"); plt.ylabel("Accuracy")
plt.title("Accuracy vs Época (sin embeddings)")
plt.grid(True, linestyle=":")
plt.legend()
plt.tight_layout()
plt.show()

plt.figure(figsize=(7,4))
plt.plot(range(1, epochs+1), train_f1_hist_c, marker="o", label="Train")
plt.plot(range(1, epochs+1), val_f1_hist_c, marker="o", label="Val")
plt.xlabel("Época"); plt.ylabel("F1 macro")
plt.title("F1 macro vs Época (sin embeddings)")
plt.grid(True, linestyle=":")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
model_c.eval()
all_preds_c, all_targets_c = [], []
with torch.no_grad():
    for x, y in val_loader_c:
        x = x.to(device)
        logits = model_c(x)
        preds = (torch.sigmoid(logits) > 0.5).float()
        all_preds_c.append(preds.cpu().numpy())
        all_targets_c.append(y.numpy())

all_preds_c = np.concatenate(all_preds_c)
all_targets_c = np.concatenate(all_targets_c)

print(classification_report(all_targets_c, all_preds_c, target_names=["<=50K", ">50K"]))

cm_abs_c = confusion_matrix(all_targets_c, all_preds_c)
cm_norm_c = confusion_matrix(all_targets_c, all_preds_c, normalize="true")

fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))
sns.heatmap(cm_abs_c, annot=True, fmt="d", cmap="Blues",
            xticklabels=["<=50K", ">50K"], yticklabels=["<=50K", ">50K"], ax=axes[0])
axes[0].set_title("Matriz de confusión (absoluta) - sin embeddings")
axes[0].set_xlabel("Predicho"); axes[0].set_ylabel("Real")

sns.heatmap(cm_norm_c, annot=True, fmt=".2f", cmap="Blues",
            xticklabels=["<=50K", ">50K"], yticklabels=["<=50K", ">50K"], ax=axes[1])
axes[1].set_title("Matriz de confusión (normalizada) - sin embeddings")
axes[1].set_xlabel("Predicho"); axes[1].set_ylabel("Real")

plt.tight_layout()
plt.show()
```

Mediante el desarrollo de los dos modelos, uno utilizando embeddings y otro que codificando con One-Hot Encoding, se pudieron observar diferencias en la performance, la cual empeora al quitar los embeddings. A continuación, se desarrollan un poco los motivos de esto:

- Al quitar los embeddings, en el modelo c) se produce una *pérdida de la representación densa* que aprenden los embeddings. Con embedding, las categorías con comportamiento similar respecto al target, por ejemplo, países con distribución de ingreso parecida, pueden terminar ubicadas cerca en el espacio vectorial aprendido, algo que utilizando One-Hot Encoding no se puede capturar. En este último, cada categoría es una dimensión aislada, sin ninguna noción de "parecido" entre categorías.

- Además, se le quita el *dropout*, el cual actúa como regularizador, que en principio podría verse como una ayuda en la métrica de train, entendiendo como una mayor capacidad efectiva de ajustarse a train. Pero se observa que con el avance de las épocas, el modelo c) muestra que las curvas de train y val se alejan entre sí.

Mientras que en el modelo c), esas misma variables entran como One-Hot Encoding completo (14+41+80=135 columnas solo entre esas tres features).

En cuanto a los parámetros, se observa que hay una diferencia importante entre el modelo b) (61839) y el modelo c) (84225). Esto no viene de los embeddings sino que de la capa 1, ya que para el modelo c) entran al MLP 167 entradas, gran parte debido al One-Hot Encoding. Para el caso b), la entrada es de 59649 y se le suman las 3 tablas de embeddings, dando un total de 61839. Acá se observa el poder de los embeddings, que son una forma genuina de representación compacta de la información, y esa compactación le ahorra alrededor de 22000 parámetros al modelo completo además de mejorar el desempeño.

Como era de esperarse, el modelo b) le gana en todas las métricas al modelo c), con una marcada diferencia sobre todo en la clase minoritaria *>50K*, lo cual sugiere que la representación densa de los embeddings es muy útil cuando hay menos datos para aprender un patrón. Además, como ya se mencionó, hay una mayor *eficiencia* en la representación logrando un mejor desempeño con una entrada al MLP más reducida.

Por último, la falta de dropout en el modelo c) y la mayor cantidad de entradas es una situación que facilita el overfitting, el cual queda evidenciado en la separación de las curvas de train y validación.
