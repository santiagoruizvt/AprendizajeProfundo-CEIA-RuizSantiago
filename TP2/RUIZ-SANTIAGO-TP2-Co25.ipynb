## a) Análisis exploratorio de los datasets

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

train = pd.read_csv("adult_train.csv")
val   = pd.read_csv("adult_val.csv")

print("Train:", train.shape, "| Val:", val.shape)
train.info()
train.describe()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(10,4))
train["income"].value_counts().plot(kind="bar", ax=ax[0], title="Train - income")
val["income"].value_counts(normalize=True).plot(kind="bar", ax=ax[1], title="Val - income (%)")
plt.tight_layout()
plt.show()

print(train["income"].value_counts(normalize=True))
```

```python
num_cols = ["age", "capital-gain", "capital-loss", "hours-per-week"]

fig, axes = plt.subplots(2, 2, figsize=(12,8))
for ax, col in zip(axes.flatten(), num_cols):
    sns.histplot(data=train, x=col, hue="income", bins=40, kde=False, ax=ax)
    ax.set_title(col)
plt.tight_layout()
plt.show()

train[num_cols].corr()
```

```python
cat_cols = ["workclass", "education", "marital-status", "occupation","relationship", "race", "sex", "skill-profile", "native-country"]

cardinality = train[cat_cols].nunique().sort_values()
print(cardinality)

# Umbral de 10 o más valores únicos => alta cardinalidad
cardinality_df = cardinality.reset_index()
cardinality_df.columns = ["variable", "n_unicos"]
cardinality_df["alta_cardinalidad"] = cardinality_df["n_unicos"] >= 10
print(cardinality_df)
```

```python
for col in cat_cols:
    plt.figure(figsize=(8, max(3, train[col].nunique()*0.3)))
    order = train[col].value_counts().index
    sns.countplot(data=train, y=col, order=order, hue="income")
    plt.title(f"{col} vs income")
    plt.tight_layout()
    plt.show()
```

```python
# Matriz de correlación

train_corr = train.copy()
train_corr["income_bin"] = (train_corr["income"] == ">50K").astype(int)

plt.figure(figsize=(6,5))
sns.heatmap(train_corr[num_cols + ["income_bin"]].corr(), annot=True, fmt=".2f",
            cmap="coolwarm", vmin=-1, vmax=1, square=True, linewidths=0.5)
plt.title("Correlación numéricas + target")
plt.tight_layout()
plt.show()
```

```python
for col in cat_cols:
    only_in_val = set(val[col].unique()) - set(train[col].unique())
    if only_in_val:
        print(f"{col}: categorías en val no vistas en train -> {only_in_val}")
if not only_in_val:
    print("No hay categorías en val no vistas en train")
```

### Análisis del EDA

- La variable `education` tiene 16 valores pero con orden lógico intrínseco, ya que cada nivel de educación es seguido por otro en un orden determinado y se puede decir que hay un orden semántico (Preschool<1st-4th<...<). Es posible realizar ordinal encoding.
- `native-country`, `occupation`, `skill-profile` son candidatas naturales a embeddings, ya que poseen alta cardinalidad sin un orden claro.
- `workclass`,`marital-status`,`relationship`,`race`,`sex` poseen baja cardinalidad y sin orden, por lo que se podría utilizar One-Hot encoding directamente.
- En la matriz de correlación, se agregó el `income` en forma binaria, mapeandolo como 0/1 para entender de forma cuantitativa qué variables numéricas pesan más en el ingreso, por ejemplo `age` o `hours-per-week`
